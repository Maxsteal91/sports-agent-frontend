```python
import os

# Configurazione: aggiungi estensioni o cartelle da ignorare
EXCLUDE_DIRS = {'node_modules', '.git', '__pycache__', 'dist', 'build', '.next', 'venv'}
ALLOWED_EXTENSIONS = {'.js', '.ts', '.jsx', '.tsx', '.py', '.json', '.env', '.html', '.css'}
OUTPUT_FILE = "contesto_progetto.txt"

def genera_contesto():
    with open(OUTPUT_FILE, 'w', encoding='utf-8') as out:
        out.write("=== STRUTTURA DEL PROGETTO ===\n")
        
        # 1. Genera l'albero delle cartelle
        for root, dirs, files in os.walk('.'):
            dirs[:] = [d for d in dirs if d not in EXCLUDE_DIRS]
            livello = root.replace('.', '').count(os.sep)
            indentazione = ' ' * 4 * livello
            out.write(f"{indentazione}[D] {os.path.basename(root)}/\n")
            sub_indenzazione = ' ' * 4 * (livello + 1)
            for f in files:
                if os.path.splitext(f)[1] in ALLOWED_EXTENSIONS:
                    out.write(f"{sub_indenzazione}[F] {f}\n")
        
        out.write("\n=== CONTENUTO DEI FILE ===\n")
        
        # 2. Legge il contenuto dei file consentiti
        for root, dirs, files in os.walk('.'):
            dirs[:] = [d for d in dirs if d not in EXCLUDE_DIRS]
            for f in files:
                estensione = os.path.splitext(f)[1]
                if estensione in ALLOWED_EXTENSIONS and f != OUTPUT_FILE and f != 'crea_contesto.py':
                    filepath = os.path.join(root, f)
                    out.write(f"\n--- INIZIO FILE: {filepath} ---\n")
                    try:
                        with open(filepath, 'r', encoding='utf-8') as file_content:
                            out.write(file_content.read())
                    except Exception as e:
                        out.write(f"[Errore nella lettura del file: {str(e)}]\n")
                    out.write(f"\n--- FINE FILE: {filepath} ---\n")

    print(f"✅ File {OUTPUT_FILE} creato con successo nella radice del progetto!")

if __name__ == "__main__":
    genera_contesto()
```
